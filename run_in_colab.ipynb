## Design your own patch

Change the frequency, laminate or thickness and read off the design.

```python
import math
from src.patch import synthesize
from src.substrates import Substrate

d = synthesize(5.8e9, Substrate('RO4350B', 3.66, 0.0037), h=0.762e-3)
print(f'W = {d.w*1e3:.2f} mm, L = {d.l*1e3:.2f} mm, inset = {d.inset_depth*1e3:.2f} mm')
print(f'Q_t = {d.q_total:.1f}, efficiency = {d.radiation_efficiency:.1%}, '
      f'VSWR<=2 bandwidth = {d.bandwidth_vswr2:.2%}')
print(f'directivity = {10*math.log10(d.directivity):.2f} dBi, gain = {10*math.log10(d.gain):.2f} dBi')
```
